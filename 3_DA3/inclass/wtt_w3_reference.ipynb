# Reference:  Function, Method & Keyword Used in Week 3

## 1. Python basics used throughout
| Name | What it does |
|---|---|
| `print(x)` | displays `x` in the cell's output |
| `type(x)` | tells you what kind of object `x` is (`str`, `dict`, `list`, `int`...) |
| `len(x)` | counts how many items are in `x` (characters in a string, rows in a dataframe, items in a list...) |
| `range(n)` | generates the sequence `0, 1, 2, ..., n-1` |
| `bool(x)` | converts `x` to `True` or `False` |
| `min(x)` / `max(x)` | smallest / largest value |
| `sum(x)` | adds up all the values in `x` |
| `f'text {x}'` | an f-string — embeds the value of `x` directly inside a piece of text |
| `del x` | deletes `x` entirely (used on a DataFrame column: `del df['col']`) |
| `x in y` | checks whether `x` is inside the collection `y` (a list, a set, dict keys...) |


## 2. Control flow & writing your own functions

| Name | What it does |
|---|---|
| `def my_func(x):` | defines your own reusable function named `my_func` |
| `if / elif / else` | runs different code depending on a condition |
| `for item in collection:` | repeats code once for every item in a list/column/etc. |
| `break` | immediately exits the current loop |
| `try: ... except SomeError as e:` | runs code that might fail, and catches the error instead of crashing |
| `assert condition, 'message'` | stops immediately with `'message'` if `condition` is `False`; does nothing if it's `True` |
| `[expr for item in collection]` | a **list comprehension** — a compact way to build a new list by transforming every item in an existing one |


## 3. Dictionaries, lists, and reading JSON

| Name | What it does |
|---|---|
| `d['key']` | gets the value for `'key'` out of a dictionary `d` |
| `d.items()` | gives you every `(key, value)` pair in a dictionary, usually inside a `for` loop |
| `lst[0]` | gets the first item of a list (indexing starts at 0) |
| `isinstance(x, dict)` | checks whether `x` is currently a dictionary (or any other type you name) |
| `json.loads(text)` | turns a JSON **string** into a real Python dictionary/list you can index into |
| `text.splitlines()` | splits one big string into a list of strings, one per line |


## 4. Regular expressions (the `re` module)

A regular expression ("regex") is a tiny pattern-matching language for text — instead of an exact string, you describe a *shape* of text to look for.

| Name | What it does |
|---|---|
| `re.findall(pattern, text)` | returns a **list** of every match of `pattern` found in `text` |
| `re.search(pattern, text)` | returns the **first** match only (or `None` if there isn't one) — good for a yes/no check |

**Patterns used in these notebooks:**

| Pattern | Means |
|---|---|
| `r'@\S+'` | an `@` followed by one or more non-space characters — matches `@username` |
| `r'#\w+'` | a `#` followed by one or more word characters (letters/digits/underscore) — matches `#hashtag` |
| `r'^RT @\w+:'` | the very **start** of the text (`^`) is `RT @someone:` — matches a retweet |
| `r'@(\w+)'` | same as `@\S+`, but the `(...)` captures just the username *without* the `@` |


## 5. Reading data into pandas

| Name | What it does |
|---|---|
| `pd.read_csv(path)` | reads a comma-separated file into a DataFrame |
| `pd.read_json(path, lines=True)` | reads a **JSONL** file (one JSON object per line) into a DataFrame — `lines=True` is required for that format, see `wtt_w3_json_and_concepts.ipynb` Part 0 |


## 6. Inspecting a DataFrame

| Name | What it does |
|---|---|
| `df.shape` | `(rows, columns)` — an **attribute**, no `()` |
| `df.columns` | the column names |
| `df.dtypes` | the data type of every column |
| `df.head(n)` | the first `n` rows (default 5) |
| `df.describe()` | count/mean/std/min/max etc. for every numeric column |
| `df['col'].isna()` | `True`/`False` per row — is this value missing? |
| `df['col'].isna().sum()` | how many missing values in that column (summing `True`=1, `False`=0) |
| `df['col'].unique()` | every distinct value that occurs, as an array |
| `df['col'].nunique()` | how many distinct values (a single number) |
| `df['col'].is_unique` | `True` if every value in the column is different (no repeats) — an **attribute** |


## 7. Selecting & filtering rows and columns

| Syntax | What it does |
|---|---|
| `df['col']` | selects ONE column, as a Series |
| `df[['col1', 'col2']]` | selects MULTIPLE columns (note the double brackets — the inner `[...]` is a list, the outer does the selecting), returns a DataFrame |
| `df[condition]` | **boolean indexing**: keeps only the rows where `condition` is `True` |
| `df['col'].iloc[0]` | selects a row **by position** (0 = first), regardless of the row's label/index |


## 8. Cleaning data

| Name | What it does |
|---|---|
| `df.dropna(subset=['a', 'b'])` | removes rows with a missing value **in those specific columns** (ignores missing values elsewhere) |
| `df.fillna(0)` | replaces missing values with `0` (or any value you give it) instead of removing the row |
| `pd.to_datetime(df['col'])` | converts a text column into real dates you can compare/sort/filter |
| `pd.to_numeric(df['col'], errors='coerce')` | converts a column to numbers; `errors='coerce'` turns anything that *can't* convert into a missing value instead of crashing |
| `df.drop_duplicates(subset=['id'])` | keeps only the first row for each unique value of `id`, drops the rest |
| `df['col'].replace(to_replace=pattern, value=x, regex=True)` | replaces every match of a regex `pattern` with `x` |
| `df['col'].replace(0, pd.NA)` | replaces one specific value (here, `0`) with another — no regex needed for an exact match |


## 9. Creating & transforming columns

| Syntax | What it does |
|---|---|
| `df['new'] = ...` | creates a new column (or overwrites an existing one) |
| `df['col'].apply(func)` | runs `func` once per **value** in the column, collects the results into a new column |
| `df.apply(func, axis=1)` | runs `func` once per **row** (`axis=1`) — `func` receives the whole row, not just one value, so it can read/combine several columns at once |
| `df['col'].str.match(pattern)` | `True`/`False` per row — does the text **start with** this regex pattern? |
| `df.explode('col')` | if a column holds a **list** per cell, turns each list item into its own row (repeating the rest of that row's data) |


## 10. Combining dataframes

| Name | What it does |
|---|---|
| `pd.concat([df1, df2, ...])` | stacks dataframes on top of each other (same columns, more rows) |
| `df1.merge(df2, on='id', how='left')` | combines two dataframes side by side, matching rows where `id` is equal |
| `left_on=..., right_on=...` | like `on=`, but for when the matching column has a **different name** in each dataframe |
| `suffixes=('_a', '_b')` | when both dataframes have a column with the same name (other than the merge key), this decides what to rename them to instead of crashing |


````python
page1 = pd.DataFrame({'id': [1, 2], 'text': ['hi', 'bye']})
page2 = pd.DataFrame({'id': [3], 'text': ['hey']})
print(pd.concat([page1, page2]))          # stacked into one dataframe

tweets = pd.DataFrame({'author_id': [1, 2, 3], 'text': ['a', 'b', 'c']})
users = pd.DataFrame({'user_id': [1, 2], 'name': ['Alice', 'Bob']})

merged = tweets.merge(users, how='left', left_on='author_id', right_on='user_id')
print(merged)   # author 3 has no matching user -> name is missing (NaN), because how='left'
````

## 11. Aggregating & summarizing

| Name | What it does |
|---|---|
| `df['col'].value_counts()` | counts how often each distinct value occurs, sorted from most to least common |
| `df.groupby('col')` | groups rows that share the same value of `col` together, so you can summarize each group separately |
| `df.groupby('col')['other_col'].mean()` (or `.std()`, `.sum()`...) | the average (or other statistic) of `other_col`, computed **separately for each group** |
| `df.transpose()` | flips rows and columns — useful when a summary table is wider than it is tall (or vice versa) and hard to read |
| `df.sort_values(ascending=False)` | sorts by value; `ascending=False` means biggest first |
| `df.sample(n, random_state=1)` | picks `n` random rows; `random_state=1` makes the "random" choice repeatable (same rows every time you run it) |


````python
demo = pd.DataFrame({
    'topic': ['climate', 'climate', 'tech', 'tech', 'tech'],
    'likes': [10, 30, 5, 15, 25],
})

print(demo['topic'].value_counts())
print(demo.groupby('topic')['likes'].mean())
print(demo.groupby('topic')['likes'].mean().transpose())
print(demo.groupby('topic')['likes'].mean().sort_values(ascending=False))
print(demo.sample(2, random_state=1))    # run this cell again - you'll get the SAME 2 rows each time
````

Some lines in the notebooks combine several of the tools from Part A in one go. Same template each time:

```
CODE:            <the exact line>
Variable(s):     what's being read from / assigned to
Function/Method: what's being called, one-line plain-English definition
Argument(s):     each argument and what it controls
Selection/Index: any [ ] column, row, or boolean selection, explained
Why we use it:   the practical problem this line solves
```


### B1. `subset=` keyword argument

```
CODE: tweets = tweets.dropna(subset=['retweet_count', 'reply_count', 'like_count', 'quote_count'])

Variable(s):     tweets - reassigned; the cleaned result overwrites the original
Function/Method: .dropna() - a DataFrame method that removes rows with missing (NaN) values
Argument(s):     subset=[...] - instead of checking ALL columns, only check these four
Selection/Index: the list itself IS the selection - it tells dropna() which columns matter
Why we use it:   we only want to drop a row because these 4 engagement counts are missing
                 - not because some unrelated column (e.g. description) happens to be empty
```


### B2. Boolean filtering (`df[condition]`)

```
CODE: missing = tweets[tweets['like_count'].isna()]

Variable(s):     missing - a new dataframe holding only the filtered rows
Function/Method: not a method call - this is boolean indexing: tweets[ ] keeps only rows
                 where the thing inside the brackets is True
Argument(s):     tweets['like_count'].isna() - a column of True/False, one per row
Selection/Index: tweets['like_count'] selects one column first; .isna() then tests each value
Why we use it:   splits off just the rows with missing values so we can inspect them
                 (e.g. see which authors they belong to) without touching the rest
```


### B3. `.apply()` with a custom function

```
CODE: tweets['topic'] = tweets['topic_raw'].apply(recategorize)

Variable(s):     creates a new column 'topic' on tweets
Function/Method: .apply() - runs a function once per value in a column, one at a time
Argument(s):     recategorize - a function WE wrote; apply() calls it automatically
                 for every row's topic_raw value and collects the results
Selection/Index: tweets['topic_raw'] selects the single column fed into the function
Why we use it:   apply() lets you use logic that isn't a built-in pandas function -
                 here, turning many specific categories into a few broad ones
```


### B4. `.merge()` with multiple keyword arguments

```
CODE: merged = tweets.merge(sentiment, on='tweet_id', how='left')

Variable(s):     merged - a brand-new dataframe combining two tables
Function/Method: .merge() - joins two dataframes into one, matching rows on a shared column
Argument(s):     sentiment - the OTHER table to combine with (tweets is "left" since
                 the method is called ON it)
                 on='tweet_id' - the shared column used to match rows between the two
                 how='left' - keep every row from tweets; attach sentiment where it
                 exists, leave it missing where it doesn't
Why we use it:   tweets and sentiment scores live in two separate files - merge() is
                 how you reconnect them using the id they have in common
```


### B5. Method chain: `groupby → select → aggregate → transpose`

```
CODE: merged.groupby('topic')[engagement_cols].mean().transpose()

Variable(s):     none assigned here - just displayed; reads from merged
Function/Method: FOUR steps chained in one line:
                 .groupby('topic')   -> bundles rows that share the same topic together
                 [engagement_cols]   -> keeps only the numeric columns we care about
                 .mean()             -> averages those columns WITHIN each topic group
                 .transpose()        -> flips rows/columns so the table reads more easily
Argument(s):     'topic' (what to group by), engagement_cols (which columns to summarize)
Why we use it:   answers "which topic gets the most likes/retweets on average?" -
                 each step narrows "all tweets" down to one number per topic per metric
```


### B6. Double-bracket column selection

```
CODE: df_min = df[['tweet_id', 'username', 'text', 'sentiment_compound',
                    'like_count', 'retweet_count', 'followers_count', 'verified']]

Variable(s):     df_min - a new, smaller dataframe
Function/Method: not a method - column selection using a LIST inside brackets
Selection/Index: outer [ ] does the selecting; inner [ ] is the list of column
                 names to keep, in that order - this is why it's double brackets
Why we use it:   keeps only the columns relevant to the research question
                 (data minimization) and drops everything else
```


### B7. Regex-based `.replace()`

```
CODE: df_min['text'] = df_min['text'].replace(to_replace=r'@\S+', value='@mention', regex=True)

Variable(s):     overwrites the existing 'text' column
Function/Method: .replace() - swaps out matching values/patterns for something else
Argument(s):     to_replace=r'@\S+' - a regular expression meaning "an @ followed by
                 one or more non-space characters" i.e. any mention
                 value='@mention' - what to put in its place
                 regex=True - treat to_replace as a PATTERN to search for, not an
                 exact literal string to match
Why we use it:   removes usernames hidden inside free text, without touching
                 the rest of the tweet
```


### B8. Mixed pandas → plain-Python indexing

```
CODE: df_jsonl['data'].iloc[0][0]

Variable(s):     reads from df_jsonl, no assignment
Function/Method: THREE different kinds of indexing stacked:
                 ['data']  -> selects the 'data' column (pandas column selection)
                 .iloc[0]  -> selects the first ROW of that column, by position (pandas)
                 [0]       -> indexes into THAT value, which is now a plain Python
                             list, not a pandas object anymore (plain Python)
Why we use it:   shows exactly where you cross from "pandas syntax" into
                 "plain Python syntax" within a single line - useful for
                 debugging when something in a nested structure looks wrong
```
