### Note: what is Machine Learning, and how is it different from statistics?

**Machine Learning (ML)**: methods that *learn patterns from data* so they can make predictions or decisions, instead of us writing the rules by hand.
Example: a spam filter is not given a list of rules; it learns from thousands of emails labelled "spam" / "not spam" which words and patterns predict spam.


| | Statistics | Machine Learning |
|---|---|---|
| Goal | **explain / test**: is there a real relationship between X and Y? | **predict**: what Y do we expect for a new case? |
| Typical question (tutorial case) | Do CPC campaigns significantly change pageviews? | How many pages will this new visitor from a CPC campaign view? |
| What we look at | coefficients and **p-values** | **predictions** (and how accurate they are) |
| Python package | `statsmodels` | `scikit-learn` (`sklearn`) |

The same regression model can serve both goals: to **test** a hypothesis you read the coefficient and its p-value (`statsmodels` summary); to **predict** you use the model's predictions (`sklearn` `.predict` / `.predict_proba`). The tutorial does exactly this with both packages.

# The different types of ML

Let's discuss the Sklearn Machine Learning Map:
https://scikit-learn.org/stable/tutorial/machine_learning_map/index.html

### Note: reading the ML map

Start at **START** and follow the questions. The four big areas of the map are:

| Area | When | Type of ML | Real-world example |
|---|---|---|---|
| **Regression** | you predict a **quantity** (a number) | supervised | predicting a house's price from its size and location; forecasting next month's sales |
| **Classification** | you predict a **category** and have **labelled** data | supervised | a spam filter (spam / not spam); predicting which customers will cancel their subscription |
| **Clustering** | you want groups but have **no labels** | unsupervised | customer segmentation: grouping shoppers with similar buying behaviour for targeted marketing |
| **Dimensionality reduction** | you want to explore or simplify many variables | unsupervised | summarising a 40-question survey into a few underlying themes; compressing images |

- **Supervised**: we know the correct answer (the label or DV) for past cases, and the model learns to predict it.
- **Unsupervised**: there is no correct answer to learn from; the model looks for structure on its own.

Note: despite its name, *logistic regression* is a **classification** model: it predicts the probability of a category.
